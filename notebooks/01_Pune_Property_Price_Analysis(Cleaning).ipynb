## Business Understanding
***Problem*** ***Statement***

The real estate market in Pune exhibits significant variation in property prices based on factors such as location, total area, configuration, amenities, and furnishing status. However, property data available across platforms is often inconsistent, incomplete, and unstructured, making it difficult for buyers to accurately estimate fair property prices or compare available options efficiently.

As a result, there is a need for a data-driven system that can clean, structure, and analyze property data to:

* Predict property prices accurately

* Provide intelligent recommendations based on user preferences

* Offer interactive insights into market trends







***Objective of Notebook***

To analyze and prepare a raw Pune property dataset for machine learning by:


*   Understanding the structure and quality of the data.
*   Identifying inconsistencies, missing values and non atomic features.
*  cleaning and Preparing the dataset for further analysis and modeling.



## Data Understanding

***Data Descripition***

The dataset Pune_property_data.csv contains 37,500+ property listings. Each row represents a single property listing, and each column describes a specific attribute related to property.

***Key Attributes Observed***
Some important columns include:



- bhk – Number of bedrooms

- bathroom – Number of bathrooms

- area / carpetarea – Property size information

- price / pricepersquare – Pricing details

- locality – Area within Pune

- projectname – Name of the residential project

- status – Ready to move / Under construction

- amenitiesavailable / amenitiesnot – Amenities information

***Cleaning age column***

```python
data["age"]
```

```python
def clean_age(x):
    if pd.isnull(x):
        return np.nan

    x = str(x).lower().replace("years", "").replace("year", "").strip()

 # case = 2-3
    if "-" in x:
      a, b = x.split("-")
      return (int(a.strip()) + int(b.strip())) / 2

# case = 2
    else:
      return int(x)

```

***Cleaning "bhk" column***

```python
data["bhk"].unique()
```

```python
# Cleaning bhk col
def clean_bhk_totalrooms(x):
  if pd.isnull(x):
    return np.isnan

  x = str(x).strip().upper().split(" ")

  if x[0].isdigit():
    rooms = int(x[0])
    t = x[1]

    if t == "BHK":
      return rooms + 2
    elif t == "RK":
      return 2

  else:
    return np.nan
```

```python
data["total rooms"] = data["bhk"].apply(clean_bhk_totalrooms)
```

```python
data[["bhk","total rooms"]].head()
```

***Cleaning "possesiondate" Column***

```python
data["possesiondate"].unique()
```

```python
# Cleaning the possesion data column
# Extracting the year column

def possesion_year(x):
  if pd.isnull(x):
    return np.nan

  x = str(x).strip().split("-")

  if x[-1].isdigit():
    return x[-1]

data["possesion year"] = data["possesiondate"].apply(possesion_year)

# Extracting the Month Column

def possesion_month(x):
  if pd.isnull(x):
    return np.nan

  x = str(x).strip().split("-")

  if x[-1].isdigit():
    return x[1]

data["possesion month"] = data["possesiondate"].apply(possesion_month)
```

***Cleaning the "status" column***

```python
data["status"].unique()
```

```python
# Extracting the "Read to move" from status col

def check_read_to_move(x):
  if pd.isnull(x):
    return np.nan

  x = str(x).lower()
  if "read to move" in x:
    return "Yes"
  else:
    return "No"

data["ready to move"] = data["status"].apply(check_read_to_move)

# Extracting the "Unfurnished" from status col

def check_unfurnished(x):
  x = str(x).lower()
  if pd.isnull(x):
    return np.nan
  elif "unfurnished" in x:
    return "Yes"
  else:
    return "No"

data["unfurnished"] = data["status"].apply(check_unfurnished)

# Extracting the "Under Construction' from status col

def check_under_construction(x):
  x = str(x).lower()
  if pd.isnull(x):
    return np.nan
  elif "under construction" in x:
    return "Yes"
  else:
    return "No"

data["under construction"] = data["status"].apply(check_under_construction)


# Extracting the "Semi furnished' from status col

def check_semi_furnished(x):
  x = str(x).lower()
  if pd.isnull(x):
    return np.nan
  elif "Semi-Furnished" in x:
    return "Yes"
  else:
    return "No"

data["semi furnished"] = data["status"].apply(check_semi_furnished)

# Extracting the "furnished" from status col

def check_furnished(x):
  x = str(x).lower()
  if pd.isnull(x):
    return np
  elif "furnished" in x:
    return "Yes"
  else:
    return "No"

data["furnished"] = data["status"].apply(check_furnished)
```
